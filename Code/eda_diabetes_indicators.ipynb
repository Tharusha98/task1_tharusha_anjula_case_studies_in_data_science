# EDA - CDC Diabetes Health Indicators

Exploratory look at `Dataset/diabetes-indicators/diabetes_binary_health_indicators_BRFSS2015.csv`.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("../Dataset/diabetes-indicators/diabetes_binary_health_indicators_BRFSS2015.csv")
print(df.shape)
print(df.isna().sum().sum(), "missing values")
df.head()
```

```python
# target is heavily imbalanced, worth keeping in mind for the train notebook
df["Diabetes_binary"].value_counts(normalize=True)
```

```python
fig, axes = plt.subplots(1, 2, figsize=(11, 4))
df["BMI"].plot(kind="hist", bins=30, ax=axes[0])
axes[0].set_xlabel("BMI")
df["Age"].value_counts().sort_index().plot(kind="bar", ax=axes[1])
axes[1].set_xlabel("age bracket")
plt.tight_layout()
```

```python
# correlation of the numeric/ordinal features with the target
corr = df.corr(numeric_only=True)["Diabetes_binary"].drop("Diabetes_binary").sort_values()
corr.plot(kind="barh", figsize=(6, 7))
plt.xlabel("correlation with Diabetes_binary")
plt.tight_layout()
```
