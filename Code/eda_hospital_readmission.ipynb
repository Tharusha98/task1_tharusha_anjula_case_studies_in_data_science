# EDA - Hospital Readmission (UCI Diabetes 130-US Hospitals)

Exploratory look at `Dataset/hospital-readmission/diabetic_data.csv` before any cleaning.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option("display.max_columns", 50)
```

```python
df = pd.read_csv("../Dataset/hospital-readmission/diabetic_data.csv")
df = df.replace("?", np.nan)
print(df.shape)
df.head()
```

```python
# how much is actually missing, and where
missing = df.isna().mean().sort_values(ascending=False)
missing[missing > 0]
```

```python
# target balance - most encounters are not readmitted within 30 days
df["readmitted"].value_counts(normalize=True)
```

```python
fig, axes = plt.subplots(1, 2, figsize=(11, 4))
df["time_in_hospital"].plot(kind="hist", bins=14, ax=axes[0])
axes[0].set_xlabel("time in hospital (days)")
df["number_inpatient"].plot(kind="hist", bins=20, ax=axes[1])
axes[1].set_xlabel("prior inpatient visits")
plt.tight_layout()
```

```python
# readmission rate by discharge disposition, top categories only
top_disp = df["discharge_disposition_id"].value_counts().nlargest(8).index
sub = df[df["discharge_disposition_id"].isin(top_disp)]
rate = sub.groupby("discharge_disposition_id")["readmitted"].apply(lambda s: (s == "<30").mean())
rate.sort_values().plot(kind="barh", figsize=(6, 4))
plt.xlabel("share readmitted <30 days")
plt.tight_layout()
```
