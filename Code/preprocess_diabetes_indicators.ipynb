# Preprocess - CDC Diabetes Health Indicators

Checks `Dataset/diabetes-indicators/diabetes_binary_health_indicators_BRFSS2015.csv` and saves `Dataset/processed/diabetes_indicators_clean.csv`. Scaling happens later in the train notebook, since the scaler needs to be fit on the training split only.

```python
import pandas as pd

df = pd.read_csv("../Dataset/diabetes-indicators/diabetes_binary_health_indicators_BRFSS2015.csv")
print("Shape:", df.shape)
print("Missing values:", df.isna().sum().sum())
print("Duplicate rows:", df.duplicated().sum())
```

```python
# BRFSS survey data comes pre-encoded (binary/ordinal), so there is no missing-value
# imputation or categorical encoding to do here, just drop exact duplicate responses
df = df.drop_duplicates()
print("Shape after dropping duplicates:", df.shape)
```

```python
df.to_csv("../Dataset/processed/diabetes_indicators_clean.csv", index=False)
print("Saved", df.shape, "to Dataset/processed/diabetes_indicators_clean.csv")
```
